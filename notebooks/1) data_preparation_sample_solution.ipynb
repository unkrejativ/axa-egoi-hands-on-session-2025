##### 📦 <span style="color: #003366; font-weight: bold;">Dataset Overview</span>
Before diving into analysis, take a moment to understand the data you’ll be working with. Here’s a detailed description of the two datasets:

🗂 <span style="color: #003366; font-weight: bold;">Dataset 1: policies.csv – Motor Insurance Policies</span>  

This dataset contains one row per insurance policy. Each row provides key characteristics of the policy and its holder. The columns are as follows  
- **IDPol:** Unique ID for each policy, used to link to severity data set
- **ClaimNb:** Number of claims that occurred for the policy
- **VehPower:** Car power, in categorical format; 1 is lowest, 8 is highest power
- **VehAge:** Car age, in years
- **VehBrand:** Car brand, in categorical format
- **VechGas:** Fuel type, in categorical format (Regular or Diesel)
- **DriverBirthDate:** The Birth Date of the Driver
- **BonusMalus:** Bonus / Malus rating, <100 means bonus, >100 means malus, in continous value format
- **Region:** Policy region in France, in categorical format
- **Density:** Density rating, number of inhabitants per km^2, in continous value format
- **Area:** Density rating of the area car driver lives in, in categorical format

    | Area Code  | Density in km^2  |
    |------------|------------------|
    | **A**  | 1-50 |
    | **B**  | 50-100 |
    | **C**  | 100-500 |
    | **D**  | 500-2000 |
    | **E**  | 2001-9850 |
    | **F**  | 10008-27000 |



🗂 <span style="color: #003366; font-weight: bold;">Dataset 2: claims.csv – Filed Claims and Damages</span>  

This dataset contains data on insurance claims. Not every policy appears here – only those that had at least one reported claim. The columns are as follows
- **IDPol:** Foreign key linking to policies.csv
- **ClaimAmount:** Number of claims that occurred for the policy
